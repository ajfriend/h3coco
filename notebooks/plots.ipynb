```python
from h3coco import cells_for_state, plot_cells
import matplotlib.pyplot as plt
import h3

cells = cells_for_state('california', 7)
cells = h3.compact_cells(cells)

fig, ax = plt.subplots(figsize=(10,10))

ax.axis('off')
plot_cells(cells, ax=ax, facecolor='blue', alpha=0.5)
plt.show()
```
