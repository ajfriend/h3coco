```python
import h3coco as coco

# Get California cells and compact them
cells = coco.cells_for_state('California', 5)
compacted = coco.compact(cells)

# Plot the co-compacted set
coco.plot_cocompact(compacted, res=5, savefig='california.png')

# Or with state outline overlay
fig, ax = plt.subplots(figsize=(12, 10))
coco.plot_state('California', ax=ax)
coco.plot_cocompact(compacted, res=5, ax=ax, savefig='california.svg')
```
